```sql
-- ============================================================
-- Create the catalog (top-level container) if it doesn't already exist
-- Create the bronze/silver/gold schemas if it doesn't already exist
-- ============================================================
CREATE CATALOG IF NOT EXISTS topcars_sa;
CREATE SCHEMA IF NOT EXISTS topcars_sa.bronze;
CREATE SCHEMA IF NOT EXISTS topcars_sa.silver;
CREATE SCHEMA IF NOT EXISTS topcars_sa.gold;
USE CATALOG topcars_sa;

-- ============================================================
-- BRONZE LAYER LOAD
-- Loads 3 raw CSV files from the /Volumes/.../raw_files/ folder
-- into bronze tables (as-is, with minimal cleanup).
-- CREATE OR REPLACE = rebuild the table each run (full refresh).
-- Column names with spaces are renamed to use underscores (_).
-- ============================================================
CREATE OR REPLACE TABLE topcars_sa.bronze.dealer_bronze AS SELECT * 
FROM read_files(
    '/Volumes/topcars_sa/bronze/raw_files/SA Cars Dealers - Sheet1.csv',
    format => 'csv',
    header => true
);

CREATE OR REPLACE TABLE topcars_sa.bronze.models_bronze AS
SELECT
  ModelID,
  Brand,
  Model,
  Segment,
  `EngineSize (L)` AS EngineSize_L,
  Fuel,
  `Price (USD)` AS Price_USD,
  `Profit (USD)` AS Profit_USD
FROM read_files(
  '/Volumes/topcars_sa/bronze/raw_files/SA Cars Models - Sheet1.csv',
  format => 'csv',
  header => true
);

CREATE OR REPLACE TABLE topcars_sa.bronze.sales_bronze AS
SELECT
  SaleID,
  Date,
  DealerID,
  ModelID,
  Quantity,
  TotalPrice,
  `Total Profit` AS Total_Profit
FROM read_files(
  '/Volumes/topcars_sa/bronze/raw_files/SA Car Sales - Sheet1.csv',
  format => 'csv',
  header => true
);



```
