```sql
-- Exchange rate hardcoded for this project. Rate as of 09 Sep 2026, sourced manually from https://wise.com/gb/currency-converter/usd-to-zar-rate
-- This is a temporary solution, in a real-world scenario, we would use an API to fetch the exchange rate dynamically.
-- 1 USD = 16.28 ZAR
DECLARE OR REPLACE exchange_rate_usd_zar DECIMAL(10,2) DEFAULT 16.28;

CREATE OR REPLACE TABLE topcars_sa.silver.models_silver AS
SELECT
    ModelID,
    Brand, 
    Model,
    Segment,
    EngineSize_L,
    Fuel,
    ROUND(Price_USD * exchange_rate_usd_zar, 2) AS Price_ZAR,
    ROUND(Profit_USD * exchange_rate_usd_zar, 2) AS Profit_ZAR
FROM topcars_sa.bronze.models_bronze;

CREATE OR REPLACE TABLE topcars_sa.silver.dealer_silver AS 
SELECT * FROM topcars_sa.bronze.dealer_bronze;
```
